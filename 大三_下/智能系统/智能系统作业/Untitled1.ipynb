```python
import numpy as np
import matplotlib.pyplot as plt  # 确保已安装matplotlib

def run():
    N = 100
    nodes, flag = node_factory(N)
    iter_classes = classify(nodes, flag, k=10)  # 修正函数名
    for classes in iter_classes:
        show_plt(classes)

def dist(v_A, v_B):
    return np.sqrt(np.power((v_A[0]-v_B[0]),2) + np.power((v_A[1]-v_B[1]),2))

def node_factory(N):
    nodes = []
    selected_flag = []
    for i in range(N):
        node = [np.random.random(), np.random.random()]
        nodes.append(node)
        selected_flag.append(0)
    return nodes, selected_flag

def sel_heads(r, nodes, flags):
    P = 0.05 * (100 / len(nodes))
    Tn = P / (1 - P * (r % (1 / P)))
    heads = []
    members = []
    n_head = 0
    rands = [np.random.random() for _ in range(len(nodes))]
    for i in range(len(nodes)):
        if flags[i] == 0:  # 修正运算符
            if rands[i] <= Tn:
                flags[i] = 1
                heads.append(nodes[i])
                n_head += 1
            else:
                members.append(nodes[i])
        else:
            members.append(nodes[i])
    return heads, members

def classify(nodes, flag, k=1):  # 修正函数名
    iter_classes = []
    for r in range(k):
        heads, members = sel_heads(r, nodes, flag)
        classes = [[] for _ in range(len(heads))]
        for i in range(len(heads)):
            classes[i].append(heads[i])
        for member in members:
            dist_min = float('inf')
            head_cla = 0
            for i, head in enumerate(heads):
                d = dist(member, head)
                if d < dist_min:
                    dist_min = d
                    head_cla = i
            classes[head_cla].append(member)
        iter_classes.append(classes)
    return iter_classes

def show_plt(classes):
    fig = plt.figure()
    ax1 = plt.gca()
    ax1.set_title('WSN1')
    plt.xlabel('X')  # 修正拼写
    plt.ylabel('Y')  # 修正拼写
    icon = ['o', '*', 'x', '+', 's']
    color = ['r', 'b', 'g', 'c', 'y', 'm']
    for i in range(len(classes)):
        if not classes[i]:  # 跳过空簇
            continue
        centor = classes[i][0]
        for point in classes[i]:
            ax1.plot([centor[0], point[0]], [centor[1], point[1]], c=color[i%6], marker=icon[i%5], alpha=0.4)  # 修正plot
    plt.show()

if __name__ == "__main__":
    run()
```

```python
import numpy as np
import matplotlib.pyplot as plt
import os

def run():
    N = 100
    nodes, flag = node_factory(N)
    iter_classes = classify(nodes, flag, k=10)

    # 创建保存图片的文件夹（使用绝对路径）
    save_dir = os.path.join(os.getcwd(), "images")
    os.makedirs(save_dir, exist_ok=True)
    print(f"图片将保存到: {save_dir}")

    # 遍历迭代并保存图片
    for idx, classes in enumerate(iter_classes):
        save_path = os.path.join(save_dir, f"cluster_iteration_{idx}.png")
        print(f"正在保存第 {idx} 次迭代的图片到: {save_path}")
        show_plt(classes, save_path)

def show_plt(classes, save_path=None):
    fig = plt.figure()
    ax1 = plt.gca()
    ax1.set_title('WSN1')
    plt.xlabel('X')
    plt.ylabel('Y')
    icon = ['o', '*', 'x', '+', 's']
    color = ['r', 'b', 'g', 'c', 'y', 'm']

    for i in range(len(classes)):
        if not classes[i]:
            continue
        centor = classes[i][0]
        for point in classes[i]:
            ax1.plot([centor[0], point[0]], [centor[1], point[1]], 
                     c=color[i % 6], marker=icon[i % 5], alpha=0.4)

    if save_path:
        plt.savefig(save_path, dpi=300, bbox_inches='tight')
        plt.close()
    else:
        plt.show()

# 其他函数保持不变（确保已修正拼写错误和逻辑问题）

if __name__ == "__main__":
    run()
```
